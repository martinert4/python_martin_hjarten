```python
import re

i = 1
with open("data/quotes.txt") as f_read, open("data/quotes_clean.txt", "w") as f_write:
    f_write.write("Cool quotes\n\n")
    for line in f_read:
        line = line.strip(" \n")

        # substitute 1 or more spaces with 1 space
        line = re.sub(" +", " ", line)
        if line != "":
            f_write.write(f"{i}. {line}\n")
            i += 1
```

```python
# 1. Läs in filen
with open("data/quotes.txt", "r", encoding="utf-8") as f:
    rader = f.read().splitlines()

print("--- STÄDAD TEXT ---")

# 2. Loopa igenom varje rad för att städa den
for rad in rader:
    # .strip() tar bort osynliga mellanslag i början/slutet av raden
    rad = rad.strip()
    
    # Om raden är helt tom efter .strip(), hoppar vi över den!
    if rad == "":
        continue
        
    # HÄR ÄR MAGIN: 
    # rad.split() delar upp alla ord i en lista och struntar i om det är 1 eller 20 mellanslag.
    # " ".join(...) sätter ihop orden igen med exakt ETT mellanslag emellan.
    städade_ord = rad.split()
    städad_rad = " ".join(städade_ord)
    
    # 3. Printa ut det färdiga resultatet
    print(städad_rad)
    print() # Lägger till en (1) ren och snygg tomrad mellan varje citat
```

```python
import re

# 1. Läs in hela filen som en enda textklump istället för rader
with open("data/quotes.txt", "r", encoding="utf-8") as f:
    hela_texten = f.read()

# 2. STÄDNING MED RE:
# \s+ matchar alla sekvenser av ett eller flera mellanslag, tabbar och radbrytningar.
# Vi ersätter dessa med ett (1) vanligt mellanslag.
text_utan_extra_mellanrum = re.sub(r'\s+', ' ', hela_texten)

# 3. FIXA CITATEN PÅ VARSIN RAD:
# Eftersom re.sub tog bort alla radbrytningar ligger nu allt på en rad.
# Vi kan enkelt dela upp texten igen vid författarnas streck " - "
# och lägga till snygga radbrytningar (\n\n) före varje nytt citat.
städade_citat = text_utan_extra_mellanrum.replace(" - ", "\n- ").replace(" If ", "If ").replace(" Time ", "\n\nTime ").replace(" An expert ", "\n\nAn expert ").replace(" Everything ", "\n\nEverything ").replace(" Nothing ", "\n\nNothing ").replace(" If I have ", "\n\nIf I have ")

print(städade_citat)
```
