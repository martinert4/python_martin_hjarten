## Try except

- try block -> testar farlig kod / input
- except block -> hanterar felaktigt input istället för att kracha programmet
- raise -> thorw exception

```python
age = input('hur gammal är du')

print(f'du är {age} år gammal')
```

```python
try:
    age = int(input("What is your age? "))

    if not 0 <= age < 125:
        raise ValueError(f"Age must be between 0 and 124, not {age}")
    print(f"You are {age} years old")
except ValueError as err:
    print(err)
```

```python
while True: 

    try:
        age = round(float(input("What is your age? ")))

        if not 0 <= age < 125:
            raise ValueError(f"Age must be between 0 and 124, not {age}")
        print(f"You are {age} years old")
        break
    except ValueError as err:
        print(err)
```

```python
while True: 

    age = input("What is your age? ")
    try:
        

        if not 0 <= age < 125:
            raise ValueError(f"Age must be between 0 and 124, not {age}")
        print(f"You are {age} years old")
        break
    except ValueError as err:
        print(err)
```

```python
while True:
    anvandar_input = input("Skriv in ett tal mellan 0 och 125: ")
    
    try:
        # 1. Försök att göra om inmatningen till ett tal
        tal = int(anvandar_input)
        
        # 2. Kontrollera om talet ligger inom det godkända intervallet
        if 0 <= tal <= 125:
            print(f"Perfekt! Du skrev det godkända talet: {tal}")
            break  # Avbryt loopen eftersom allt gick bra
        else:
            print("Fel: Talet måste vara mellan 0 och 125!")
            
    except ValueError:
        # 3. Hit hamnar programmet om användaren skrev bokstäver eller tecken
        print("Fel: Du måste skriva en siffra, inte bokstäver!")
```

```python
try:
    # Här lägger vi koden som KAN krascha
    nummer = int(input("Skriv ett tal att dela 10 med: "))
    resultat = 10 / nummer
    print(f"Resultatet blev: {resultat}")

except ValueError:
    # Denna körs om användaren skriver bokstäver (t.ex. "hej")
    print("Ojdå! Du måste skriva en siffra, inte bokstäver.")

except ZeroDivisionError:
    # Denna körs om användaren skriver siffran 0 (man kan inte dela med 0)
    print("Matematisk omöjlighet! Du kan inte dela med noll.")
```
