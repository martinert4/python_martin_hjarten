```python
# greeter - function name
# name - parameter 1
# age - parameter 2
def greeter(name, age):
    print(f"Hi {name}, you are {age} years")

# call function greeter with positional arguments "Bob" and 15
greeter("Bob", 15)
greeter("Anna", 13)
```

```python
def get_floor(floor_number):
    try:
        floor = floor_number[0]

        if floor == "1":
            print("You are on floor 1.")
        elif floor == "2":
            print("You are on floor 2.")
        elif floor == "3":
            print("You are on floor 3")
    except ValueError as err:
        print(err)


get_floor(str(207))
```

```python
def med(*nummer):
    summa = 0
    for n in nummer:
        summa = summa + n
    return summa/len(nummer) 
 
med(2,2,3,4)
```

```python
import matplotlib.pyplot as plt 

x = range(5)
y = [2,-3,5,2,8]

plt.plot(x,y)
plt.grid()
plt.title("Cool stock")
plt.xlabel(xlabel="days")
plt.ylabel(ylabel="price")
```

```python
def stock_plot(x,y, title, **options):
    print(f"Options used: {options}")
    plt.plot(x,y, **options)
    plt.grid()
    plt.title(label=title) 
    plt.xlabel(xlabel="days")
    plt.ylabel(ylabel="price")

x = range(5)
y = [-5,-30,5,20,8]

stock_plot(x,y, title="Cooler plot", linewidth=4, linestyle="--")
```

```python
stock_plot(x,y, title="hej hej plot", linewidth=2, linestyle="--")
```
