```python
# Data lagras i ett dictionary
prov_1 = {"id": "Sjö-A", "massa_g": 50.0, "volym_ml": 25.0}


# Funktionen är helt fristående från datan
def rakna_densitet(prov):
    return prov["massa_g"] / prov["volym_ml"]


# Användning
densitet = rakna_densitet(prov_1)
print(f"Densitet: {densitet} g/ml")
```

```python
class LabProver:
    # __init__ körs när vi skapar ett nytt prov (objekt)
    def __init__(self, prov_id, massa_g, volym_ml):
        self.id = prov_id  # Egenskap (Attribut)
        self.massa = massa_g  # Egenskap (Attribut)
        self.volym = volym_ml  # Egenskap (Attribut)

    # En funktion inuti en klass kallas för en metod
    def rakna_densitet(self):
        return self.massa / self.volym


# Vi skapar ett faktiskt objekt (en instans av klassen)
prov_1 = LabProver("Sjö-A", 50.0, 25.0)
prov_2 = LabProver("Sjö-B", 60.0, 45.0)
# Nu äger objektet sitt eget beteende!
#print(f"Densitet: {prov_1.rakna_densitet()} g/ml")
prov_2.rakna_densitet
```
