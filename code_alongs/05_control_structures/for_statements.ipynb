lopa genom listor

```python
djur = ['apa', 'katt', 'hund']

for djur in djur:
    print(f'söta djur: {djur}')
```

```python
djur = ['apa', 'katt', 'hund']

for i in range(1,4):
    print(f'{i}, {djur[i-1]}')
```

nästlade for

```python
for row in range(3):
    for col in range(3): 
        print([row,col], end='  ')
    print()
```
