```python
#exam resultat
# a 90
# b 80
# c 70
# d 60
# e 50
# f <50

provresultat = int(input("sätt in ditt resultat"))

if provresultat >= 90:
    print("Du får A")
elif provresultat >= 80:
    print('Du får B')
elif provresultat >= 70:
    print('Du får C')
elif provresultat >= 60:
    print('Du får D')
elif provresultat >= 50:
    print('Du får E')
else:
        print('Du är underkänd')
```
