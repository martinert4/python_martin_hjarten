# Grundkurs: Avancerad List Comprehension, Matriser & Tuples 🚀

Här går vi igenom hur du skapar mönsterbaserade listor, bygger tvådimensionella matriser (t.ex. schackbräden) samt hur `tuples` (par) fungerar och används.

---

## 🎯 Del 1: Lösa uppgiften (Skapa ['A1', 'B1', ..., 'H1'])

För att skapa listan `['A1', 'B1', 'C1', 'D1', 'E1', 'F1', 'G1', 'H1']` behöver vi loopa igenom bokstäverna A till H, och lägga till textsträngen `"1"` efter varje bokstav.

### Koden:
```python
# Vi loopar igenom en textsträng med alla bokstäver
bokstaver = "ABCDEFGH"

# List comprehension: Skapa kombinationen för varje bokstav
schack_rad = [bokstav + "1" for bokstav in bokstaver]

print(schack_rad)
# Output: ['A1', 'B1', 'C1', 'D1', 'E1', 'F1', 'G1', 'H1']
```

### 💡 Pedagogisk förklaring:
* `for bokstav in bokstaver`: Python plockar ut en bokstav i taget (`'A'`, sen `'B'`, sen `'C'`...).
* `bokstav + "1"`: Detta är instruktionen längst till vänster. Den säger: *"Ta bokstaven vi just plockade ut och klistra ihop den med texten '1'"*. 
* Allt sparas automatiskt i den nya listan!

---

## 🏁 Del 2: Gå vidare till Matriser (Nästlad List Comprehension)

Tänk om vi vill skapa ett **helt schackbräde**? Alltså `['A1', 'B1'...]`, sen `['A2', 'B2'...]` ända upp till rad 8. En lista som innehåller andra listor kallas för en **matris** (eller en tvådimensionell lista).

Vi kan göra detta genom att lägga en loop inuti en annan loop (nästling) på en enda rad!

### Koden för ett helt schackbräde (8x8 matris):
```python
bokstaver = "ABCDEFGH"

# Yttre loopen kör siffror (1 till 8). Inre loopen kör bokstäver (A till H).
schackbrade = [[b + str(s) for b in bokstaver] for s in range(1, 9)]

# Skriv ut rad för rad så det syns snyggt
for rad in schackbrade:
    print(rad)
```

### 🖨️ Output i terminalen:
```text
['A1', 'B1', 'C1', 'D1', 'E1', 'F1', 'G1', 'H1']
['A2', 'B2', 'C2', 'D2', 'E2', 'F2', 'G2', 'H2']
...
['A8', 'B8', 'C8', 'D8', 'E8', 'F8', 'G8', 'H8']
```

### 🔍 Hur läser man koden?
Läs den **utifrån och in**:
1. `for s in range(1, 9)`: "Skapa 8 rader (siffrorna 1 till 8)..."
2. `[b + str(s) for b in bokstaver]`: "...och för *varje* sådan rad, skapa en lista där alla bokstäver paras ihop med den radens siffra." *(Vi använder `str(s)` eftersom vi måste göra om siffran till text innan vi klistrar ihop den med bokstaven)*.

## 🔒 Del 3: Vad är en Tuple och hur fungerar den?

En **Tuple** (en datatyp som kallas *sträng* eller *multipel* på svenska) liknar en lista väldigt mycket, men med **en gigantisk skillnad**: 

> En tuple är **immutable** (oföränderlig). När du väl har skapat den kan du *inte* ändra på värdena, lägga till nya eller ta bort saker.

### Hur ser de ut?
Medan listor använder fyrkantsparenteser `[ ]`, använder tuples vanliga parenteser `( )`.

```python
# En lista går att ändra
min_lista = [10, 20, 30]
min_lista[0] = 99 # Fungerar utmärkt!

# En tuple är låst
min_tuple = (10, 20, 30)
# min_tuple[0] = 99 # ❌ DETTA KRASCHAR! Python tillåter det inte.
```

### Varför använder man Tuples istället för Listor?
1. **Säkerhet:** Om du har data som *aldrig* får ändras av misstag under programmets gång (t.ex. koordinater `(x, y)`, RGB-färgkoder `(255, 0, 0)` eller veckans sju dagar).
2. **Prestanda:** Tuples är mycket snabbare för datorn att hantera än listor eftersom datorn vet exakt hur stora de är och att de aldrig ändras.

### Tuple Comprehension? (Varning! ⚠️)
Om du försöker skriva en list comprehension med vanliga parenteser för att skapa en tuple, till exempel `min_tuple = (x for x in range(5))`, så skapar Python något som kallas för en *Generator* (en sorts minnessnål loopmotor), inte en tuple.

För att göra en "Tuple Comprehension" måste du därför skriva `tuple()` runt om en vanlig comprehension:

```python
# Skapar en tuple med kvadraten av talen 1 till 5
mina_kvadrater = tuple(x**2 for x in range(1, 6))

print(mina_kvadrater)
# Output: (1, 4, 9, 16, 25)  <-- Lägg märke till de runda parenteserna!
```

4. Dice rolls convergence (**)

````python
import random



resultat =[]

försök = [10, 100, 1000, 10000, 100000, 1000000]

for i in range(1,7):
    tärnings_kast = []

    for n in range(1,försök[i-1]+1):
        tärnings_kast.append(random.randint(1,6))
    antal_6 = tärnings_kast.count(6)
    enskilt_försök = försök[i-1]
    sannolikhet_6 = (antal_6/ enskilt_försök)*100
    
    resultat.append(sannolikhet_6)

resultat

````

````python
import matplotlib.pyplot as plt


# 1. Skapa data (Här är dina försök som x-värden)
x_varden = [10, 100, 1000, 10000, 100000, 1000000]
y_varden = resultat # Exempel på list comprehension för y = x^2

# 2. Skapa plottningen
plt.plot(x_varden, y_varden, marker='o', linestyle='-', color='b')

# GÖR X-AXELN BRA: Sätt logaritmisk skala eftersom värdena växer exponentiellt
plt.xscale('log') 

# Tvinga axeln att visa exakt dina mätpunkter som etiketter (ticks)
plt.xticks(x_varden, labels=['10', '100', '1 000', '10 000', '100 000', '1 000 000'])

# 3. Lägg till snygga rubriker och axelnamn
plt.title("Visualisering av $y = x^2$ per antal försök")
plt.xlabel("Antal försök (Logaritmisk skala)")
plt.ylabel("y-värde ($y = x^2$)")

# 4. Lägg till ett rutnät (grid) så att det är lättare att läsa av värdena
plt.grid(True, which="both", linestyle='--', alpha=0.6) # "both" visar rutnät för log-skalan också

# 5. Visa fönstret med diagrammet
plt.show()
````

````python
import random
import numpy as np
import math
import matplotlib.pyplot as plt


antal_sim = 5000

x_punkter =  [random.uniform(-1,1) for n in range(antal_sim)]
y_punkter =  [random.uniform(-1,1) for n in range(antal_sim)]

x_punkter_större_1 = []
y_punkter_större_1 = []

x_punkter_mindre_1 = []
y_punkter_mindre_1 = []

#c=[]

for i in range(antal_sim):
    c = math.sqrt(x_punkter[i]**2 + y_punkter[i]**2) 
    if c > 1:
        x_punkter_större_1.append(x_punkter[i])
        y_punkter_större_1.append(y_punkter[i])
    else:
            x_punkter_mindre_1.append(x_punkter[i])
            y_punkter_mindre_1.append(y_punkter[i])


antal_inner_punkter = len(x_punkter_mindre_1)
antal_yttre_punkter = len(x_punkter_större_1)

andel = antal_inner_punkter / antal_yttre_punkter


#print(x_punkter_större_1)
#print(y_punkter_större_1)
#print('')
#print(x_punkter_mindre_1)
#print(y_punkter_mindre_1)
#filt_storre_1 = [x for x in c if x > 1]
#filt_mindre_1 = [x for x in c if x <= 1]

# 1. Rita de inre punkterna (t.ex. blå)
plt.scatter(x_punkter_mindre_1, y_punkter_mindre_1, color='blue', label='Inuti (d < 1)')

# 2. Rita de yttre punkterna (t.ex. röd)
plt.scatter(x_punkter_större_1, y_punkter_större_1, color='red', label='Utanför (d > 1)')

# 3. Lägg till lite extra fix så grafen blir snygg och tydlig
plt.title("Monte Carlo-simulering")
plt.xlabel("X-axel")
plt.ylabel("Y-axel")
plt.legend()  # Visar rutan som förklarar färgerna (label)

# 4. Visa grafen!
plt.show()

print(andel)
````

# 7. Teori (Theory)

### a) Vad är en sekvens i Python? Ge exempel på sekvenstyper.
En **sekvens** är en ordnad samling av element där varje sak har en bestämd plats (ett index). Python kommer ihåg den ordning som elementen lades till i, vilket gör att du kan plocka ut dem baserat på deras position.
* **Exempel på sekvenstyper:** `list` (listor), `tuple` (tupler), `str` (textsträngar) och `range` (nummerföljder).

### b) Vad är skillnaden mellan en lista och en tupel?
* **Lista ``:** Är **föränderlig** (mutable). Du kan lägga till, ta bort eller ändra element i listan efter att den har skapats.
* **Tupel `(1, 2, 3)`:** Är **oföränderlig** (immutable). När den väl har skapats kan du inte ändra dess innehåll eller storlek.

### c) Hur fungerar indexering i Pythons sekvenser? Vad händer om du använder ett negativt index?
Indexering används för att komma åt ett specifikt element i en sekvens genom att ange dess positionsnummer. Python börjar alltid räkna från `0` för det första elementet.
* **Negativt index:** Gör att Python börjar räkna **bakifrån** från slutet av sekvensen. Till exempel ger `[-1]` det allra sista elementet, `[-2]` ger det näst sista, och så vidare.

### d) Hur hanterar Python index som är utanför gränserna (out-of-range) vid slicing vs indexering?
* **Indexering:** Om du försöker hämta en specifik plats som inte finns (t.ex. plats 10 i en lista med bara 3 saker) kraschar programmet och ger ett **`IndexError`**.
* **Slicing:** Python är väldigt förlåtande. Om ditt intervall går utanför sekvensens gränser (t.ex. `lista[0:50]` på en kort lista) kraschar det inte, utan returnerar bara de element som faktiskt finns tillgängliga.

### e) Vilka av följande sekvenstyper är föränderliga (mutable): list, tuple, str, range?
* **Föränderlig (Mutable):** `list`
* **Oföränderliga (Immutable):** `tuple`, `str`, `range`

### f) Vad är sekvensuppackning (sequence unpacking)? Ge ett exempel.
Sekvensuppackning innebär att du plockar ut elementen ur en sekvens (t.ex. en lista eller tupel) och tilldelar dem till flera separata variabler samtidigt på en och samma rad.

### g) Vad är några vanliga användningsområden för enumerate() och zip() när man arbetar med sekvenser?
* **`enumerate()`:** Används i loopar när du behöver hålla koll på **både elementet och dess index (position)** samtidigt (t.ex. när du vill numrera en utskriven lista).
* **`zip()`:** Används för att loopa igenom **flera sekvenser samtidigt** genom att para ihop deras element två och två, precis som ett blixtlås (t.ex. para ihop en lista med användarnamn med en lista med e-postadresser).

### h) Vad är skillnaden mellan sorted() och .sort() på listor?
* **`.sort()`:** Ändrar direkt i din **original-lista** (sorterar den på plats) och sparar inte den gamla ordningen.
* **`sorted()`:** Lämnar original-listan helt orörd och skapar istället en **helt ny sorterad kopia** som en ny lista.

````python
# AI genererad lösning

import random
import matplotlib.pyplot as plt

vinst_ratio_lista = []
antal_försök_lista = [10, 100, 1000, 10000, 100000, 1000000]

# Vi simulerar strategin att ALLTID BYTA dörr
for antal_försök in antal_försök_lista:
    antal_vinst = 0

    for _ in range(antal_försök):
        # STEG 1: Placera ut vinsten ('kanin') bakom en slumpmässig dörr (0, 1 eller 2)
        dörrar = ['orm', 'orm', 'orm']
        vinst_index = random.randint(0, 2)
        dörrar[vinst_index] = 'kanin'

        # STEG 2: Spelaren väljer en dörr slumpmässigt
        spelarens_val = random.randint(0, 2)

        # STEG 3: Programledaren öppnar en dörr som:
        # 1. Inte är spelarens dörr
        # 2. Inte innehåller kaninen
        möjliga_dörrar_att_öppna = []
        for i in range(3):
            if i != spelarens_val and dörrar[i] == 'orm':
                möjliga_dörrar_att_öppna.append(i)
        
        programledarens_val = random.choice(möjliga_dörrar_att_öppna)

        # STEG 4: Spelaren BYTER till den återstående stängda dörren
        # Den nya dörren är den som varken är spelarens första val eller programledarens öppnade dörr
        for i in range(3):
            if i != spelarens_val and i != programledarens_val:
                nytt_val = i
                break

        # STEG 5: Kolla om det blev vinst efter bytet
        if dörrar[nytt_val] == 'kanin':
            antal_vinst += 1

    # Räkna ut och spara vinst-ratio för detta antal försök
    vinst_ratio = antal_vinst / antal_försök
    vinst_ratio_lista.append(vinst_ratio)

print("Vinstchanser vid byte:", vinst_ratio_lista)

# Rita grafen (din snygga matplotlib-kod behålls!)
plt.plot(antal_försök_lista, vinst_ratio_lista, marker='o', color='blue', label='Byta dörr')
plt.xscale('log')
plt.axhline(y=2/3, color='r', linestyle='--', label='Teoretisk chans (66.7%)') # La till en mållinje
plt.title("Monty Hall-simulering (Monte Carlo)")
plt.xlabel("Antal simuleringar (Logaritmisk skala)")
plt.ylabel("Andel vinster (Vinst-ratio)")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.show()
````
