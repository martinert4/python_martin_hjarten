# 10. Theory

## a) When is it better to use while statement and when is it better to use for statement
* **Använd `for`-loopar** när du på förhand vet **hur många gånger** loopen ska köras. Det används oftast för att gå igenom en färdig samling av saker, till exempel en lista med namn, en textsträng eller ett specifikt intervall av tal (med `range()`).
* **Använd `while`-loopar** när du **inte vet** hur många gånger loopen behöver köras, utan den ska fortsätta så länge ett visst **villkor är sant**. Det är perfekt för situationer där du väntar på en specifik händelse, till exempel tills en användare skriver "avsluta" eller tills ett spel är över.

## b) What is the difference between = and == in Python?
* **Enkelt likhetstecken (`=`)** används för **tilldelning** (assignment). Det betyder att du sparar ett värde i en variabel. 
  * *Exempel:* `x = 5` betyder "ge variabeln x värdet 5".
* **Dubbelt likhetstecken (`==`)** används för **jämförelse** (comparison). Det ställer en fråga och kollar om två saker är exakt lika värda. Svaret blir alltid antingen `True` (sant) eller `False` (falskt).
  * *Exempel:* `x == 5` betyder "är x lika med 5?".

---

## c) Explain how an if-elif-else block works in Python. What happens if multiple conditions are true?
Ett `if-elif-else`-block fungerar som en beslutskedja som Python läser **uppifrån och ned**:
1. Python kollar först `if`-villkoret. Om det är sant körs den koden, och resten av blocket hoppas över.
2. Om det är falskt går Python vidare till nästa `elif` (kort för "else if") och testar det villkoret.
3. Om inga `if`- eller `elif`-villkor är sanna, körs koden under `else` som en sista utväg (en "fånga upp allt"-lösning).

**Vad händer om flera villkor är sanna?**
Python kör **bara det allra första villkoret som är sant**. Så fort Python hittar ett villkor som stämmer, körs den tillhörande koden och sedan **avbryts hela blocket**. Eventuella sanna villkor som ligger längre ner i kedjan kommer helt att ignoreras.

## d) What does the continue statement do in a loop? How is it different from break?
Både `continue` och `break` ändrar hur en loop beter sig, men på helt olika sätt:
* **`continue` hoppar över resten av den nuvarande rundan** (iterationen) i loopen och hoppar direkt till nästa runda. Loopen fortsätter alltså att köra.
* **`break` avbryter hela loopen helt och hållet**. Python hoppar omedelbart ut ur loopen och fortsätter med koden som ligger efter loopen.

*Tänk på det så här:* `continue` är som att hoppa över en jobbig uppgift i en läxa och gå till nästa uppgift. `break` är som att stänga igen matteboken och sluta plugga för dagen.

## e) Explain the meaning of truthy and falsy values and give a few example of both
I Python kan alla typer av värden (inte bara `True` och `False`) utvärderas i en `if`-sats för att se om de räknas som "sanna" eller "falska".
* **Falsy** är värden som Python ser som "tomma" eller "noll". Om du sätter dem i ett villkor räknas de som `False`.
* **Truthy** är i princip allt annat. Om ett värde innehåller något eller inte är noll, räknas det som `True`.

**Exempel på Falsy-värden:**
* `0` (siffran noll)
* `""` (en helt tom textsträng)
* `[]` (en tom lista)
* `None` (avsaknad av värde)

**Exempel på Truthy-värden:**
* `1` eller `-5` (alla siffror utom noll)
* `"Hej"` (en textsträng med innehåll)
* `[1, 2, 3]` (en lista med element)

## f) Can you nest control structures in Python (e.g., a loop inside an if)? Give a brief example.
**Ja, det går jättebra.** Du kan lägga vilken kontrollstruktur som helst inuti en annan (loopar i if-satser, if-satser i loopar, eller loopar i loopar). Det kallas för att "nästla" (nesting).

Här är ett kort exempel där en `for`-loop ligger inuti en `if`-sats:

```python
regnigt = False

if not regnigt:
    print("Det regnar inte! Vi räknar till tre utomhus:")
    for i in range(1, 4):
        print(i)
```

## g) Every expression can be part of a statement, but not all statements are expressions. True or false?
Svaret är **True** (Sant).

* **Ett uttryck (expression)** är kod som alltid **beräknas till ett värde** (t.ex. `3 + 5` blir `8`, eller `x == 5` blir `True`). Eftersom de skapar ett värde kan de alltid vara en del av en större instruktion (statement).
* **En sats (statement)** är en **hel instruktion eller handling** som Python utför (t.ex. att skapa en variabel `x = 10`, eller att starta en loop `for i in range(5):`). Dessa handlingar blir inte ett enskilt värde i sig själva. Du kan till exempel inte skriva `y = (x = 10)`, eftersom `x = 10` är en sats och inte ett uttryck som ger ett värde.
