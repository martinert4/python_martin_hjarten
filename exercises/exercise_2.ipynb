5. Arithmetic sum (*)

```python
lista = []
for n in range(1,101):
    lista.append(n)
print(sum(lista))
```

```python
lista = []
for n in range(1,100,2):
    lista.append(n)
print(sum(lista))
```

```python
lista = []
n=1
while n <= 100:

    lista.append(n)
    n=n+1


print(sum(lista))
```
