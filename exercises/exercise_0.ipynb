## 2. Theoretical exercises

### a) What is the difference between git and github?
**Svar:** 
* **Git** är själva verktyget (programvaran) som körs lokalt på din egen dator. Det håller reda på historiken och alla ändringar man gör i mina filer.
* **GitHub** är en webbplats och en molntjänst på internet där man kan ladda upp (pusha) sina Git-projekt. Man gör en säkerhetskopia i molnet och enkelt kan dela sin kod med andra eller framtida arbetsgivare.

### b) What is the purpose of .gitignore file?
**Svar:** 
Syftet med en .gitignore-fil är att berätta för Git vilka filer och mappar som programmet ska strunta i (ignorera). Man vill inte ladda upp allt till GitHub. Exempel på saker man vill ignorera är din virtuella miljö (`.venv`), hemliga lösenord/nycklar eller tillfälliga systemfiler som skapas automatiskt.

### c) Why is it a good idea to work with git and github?
**Svar:** 
Det är en jättebra idé av flera anledningar:
* **Tidsmaskin:** Git sparar  historik. Om man råkar radera något eller om koden slutar fungera kan man enkelt hoppa tillbaka till en fungerande version.
* **Säkerhet:** Tack vare GitHub har man alltid en backup i molnet om din dator skulle gå sönder.
* **Portfolio:** GitHub fungerar som ditt digitala CV där man kan visa upp vad man kan för framtida arbetsgivare.
* **Samarbete:** Det gör det supersmidigt att jobba i team, eftersom flera personer kan bygga på samma projekt samtidigt utan att skriva över varandras kod.


## 3. Git-begrepp (Glosor)

### Repository (Repo)
**Svar:** 
Ett repository (ofta kallat "repo") är helt enkelt din projektmapp, fast med en inbyggd tidsmaskin. Det är platsen där Git sparar alla mina filer tillsammans med hela historiken över alla ändringar man någonsin har gjort. Det kan ligga lokalt på din dator eller i molnet på GitHub.

### Working Directory
**Svar:** 
Detta är den vanliga mappen på din dator där man sitter och jobbar just nu. Det är här man skapar, tar bort eller ändrar i mina kodfiler och Markdown-dokument innan man har sparat dem i Gits historik.

### Staging Area (Index)
**Svar:** 
Tänk dig att man ska ta ett gruppfoto.man ropar dit tre kompisar och ber dem ställa sig på en rad och le.De står nu redo framför kameran. De har inte fastnat på bild än, men man har valt ut dem och ställt upp dem precis som man vill ha dem.I datorn: Att lägga till filer i Staging Area (när man klickar på + i VS Code) är exakt samma sak. man väljer ut precis vilka filer och ändringar som ska få vara med på nästa "bild".

Hela kedjan sammanfattad:Explorer: man skriver sina svar (filen ändras).Klickar på +: Filen flyttas till Staging Area (man förbereder paketet).Klickar på Commit: man sparar paketet i tidsmaskinen på din egen dator.Klicka på Sync Changes / Push: man skickar paketet till GitHub i molnet

### Commit
**Svar:** 
När man nöjd med hur alla står. man trycker på knappen – KLICK!Nu har man tagit bilden. Bilden sparas i kamerans minne och kan aldrig ändras i efterhand. man skriver kanske en liten text under bilden i fotoalbumet: "Kompisgänget i parken".I datorn: När man klickar på Commit trycker man på kameraknappen. Git tar ett digitalt foto av filerna som stod i Staging Area och sparar det permanent i din dators minne tillsammans med ditt meddelande.

### Push
**Svar:** 
Att "pusha" betyder att man skickar (laddar upp) sina lokala commits från din egen dator upp till ditt repository i molnet på GitHub. Det är först när man pushar som andra kan se mina ändringar och man får en säker backup på internet.
