# 9. Teori och Glosor (Funktioner & Dictionaries)

## Teori

### a) Vad är en funktion i Python, och varför används den?
En funktion är ett självständigt, namngivet kodblock som har sparats som ett återanvändbart **recept**. Funktionen körs bara när man anropar den.

**Varför den används:**
* **Minskar upprepning:** Istället för att skriva samma kod 10 gånger (copy-paste), skriver du den en gång i en funktion och anropar den 10 gånger.
* **Bättre ordning:** Koden blir lättare att läsa, förstå och felsöka när den är uppdelad i mindre, logiska bitar.
* **Enklare underhåll:** Om du behöver ändra hur en beräkning görs, behöver du bara ändra på ett enda ställe (inuti funktionen).

---

### b) Förklara skillnaden mellan parametrar (parameters) och argument (arguments).
Dessa två blandas ofta ihop, men skillnaden är om vi pratar om *receptet* eller när vi faktiskt *lagar maten*:

* **Parametrar (Mottagare):** Variablerna som vi skriver i funktionens definition (inom parenteserna när vi skapar funktionen). De fungerar som tomma platsbärare.
  * *Exempel:* I `def hej(namn):` är `namn` en parameter.
* **Argument (Inskickade värden):** De faktiska värdena som vi skickar in i funktionen när vi anropar den.
  * *Exempel:* I `hej("Pikachu")` är `"Pikachu"` argumentet.

---

### c) Vad är skillnaden mellan `*args` och `**kwargs` i en funktionsdefinition?
Båda används när vi inte vet i förväg exakt hur många argument som användaren kommer att skicka in i funktionen:

* **`*args` (Variabla positionsargument):** Tar emot extra argument som en **tuple (en låst lista)**. Används för vanliga värden på rad.
  * *Exempel:* `def plussa(*tal):` gör att du kan anropa `plussa(1, 2, 3, 4, 5)`.
* **`**kwargs` (Variabla nyckelordsargument):** Tar emot extra argument som en **dictionary**. Används när argumenten skickas in med ett namn/nyckelord.
  * *Exempel:* `def info(**data):` gör att du kan anropa `info(namn="Kalle", ålder=25)`.

---

### d) Vad är skillnaden mellan en vanlig funktion och en lambda-funktion i Python?
* **Vanlig funktion:** Skapas med nyckelordet `def`, har ett namn, kan innehålla många rader kod, dokumentation och komplex logik.
* **Lambda-funktion:** En anonym (namnlös) **minifunktion** som skrivs på **en enda rad**. Den har ingen `return`-rad eftersom den automatiskt skickar tillbaka det som räknas ut.
  * *Exempel på lambda:* `dubbla = lambda x: x * 2` (Gör exakt samma sak som en vanlig `def`-funktion, men är gjord för snabba, enkla beräkningar).

---

### e) Förklara syftet med metoden `get()`. Hur skiljer den sig från att hämta en nyckel direkt?
Båda används för hämta ett värde från en dictionary baserat på en nyckel, men de hanterar saknade nycklar helt olika:

* **Hämta direkt (`dict[nyckel]`):** Om du söker efter en nyckel som *inte finns* så kraschar programmet omedelbart med ett `KeyError`.
* **Använda `.get()`-metoden (`dict.get(nyckel)`):** Om nyckeln inte finns så kraschar inte programmet. Istället svarar Python snällt `None` (ingenting). Du kan dessutom bestämma ett eget reservsvar om nyckeln saknas, till exempel: `pokedex.get("Okänd", "Hittades inte")`.

---

### f) Vad är skillnaden mellan metoderna `.keys()`, `.values()`, och `.items()`?
Dessa tre metoder används för att plocka ut olika delar ur en dictionary:

* **`.keys()`**: Hämtar **bara alla nycklar** (t.ex. alla Pokémon-namn: `['Pikachu', 'Gengar']`).
* **`.values()`**: Hämtar **bara alla värden** (t.ex. alla typer/index: `['Elektrisk, 25', 'Spöke/Gift, 94']`).
* **`.items()`**: Hämtar **både nyckel och värde tillsammans** i par. Perfekt när man vill loopa igenom hela databasen med en `for`-loop.

---

### g) Kan nycklar i en dictionary vara av muterbara (föränderliga) typer som listor eller andra dictionaries? Varför eller varför inte?
**Nej, absolut inte.** Nycklar i en dictionary måste vara *immuterbara* (oföränderliga), som till exempel textsträngar, heltal eller tuples.

**Varför:** Python använder en supersnabb sökteknik som heter *hashing* för att hitta nycklar. Denna teknik kräver att nyckelns värde är helt låst och aldrig kan ändras. Om du skulle använda en lista som nyckel, och du sedan ändrar i listan (`.append()`), så skulle Python tappa bort var i minnet datan ligger och din dictionary skulle gå sönder.

## Gloslista

| Terminologi (Term) | Förklaring (På svenska) |
| :--- | :--- |
| **def** | Det nyckelord som används för att definiera (skapa) en ny funktion i Python. |
| **function call** | Funktionsanrop. När man faktiskt kör funktionen genom att skriva dess namn följt av parenteser, t.ex. `print()`. |
| **argument** | De faktiska värden eller variabler som skickas in i en funktion vid ett anrop. |
| **return statement** | Return-sats. Kommandot `return` som avslutar funktionen och skickar tillbaka ett resultat/svar till den som anropade den. |
| **parameter** | Platsbärare/variabler i funktionens parentes som bestämmer vilken data funktionen vill ta emot. |
| **default parameter** | Standardparameter. Ett förvalt värde som en parameter får om användaren struntar i att skicka med ett argument (t.ex. `def hej(namn="Kompis"):`). |
| **keyword arguments** | Nyckelordsargument. När man anger parameterns namn vid anropet så att ordningen inte spelar någon roll, t.ex. `avstånd(x1=0, y1=0, x2=5, y2=5)`. |
| **variable arguments** | Variabla argument (`*args`). Gör att en funktion kan ta emot hur många extra positionsargument som helst (packas ihop som en tuple). |
| **positional arguments** | Positionsargument. Vanliga argument där ordningen man skickar in dem avgör vilken parameter de hamnar i. |
| **lambda functions** | Anonyma minifunktioner som skrivs på en enda rad för enkla, snabba matematiska beräkningar eller sorteringar. |
| **dict()** | Pythons inbyggda funktion för att skapa en helt ny, tom dictionary eller göra om annan kompatibel data till en dictionary. |
| **get()** | En säker metod för att hämta värden ur en dictionary utan risk för att programmet kraschar om nyckeln saknas. |
| **items()** | En metod som returnerar alla nycklar och värden i en dictionary som par, perfekt att använda i `for`-loopar. |
| **in operator** | `in`-operatorn. Används för att snabbt kontrollera om en specifik nyckel finns inuti en dictionary (returnerar `True` eller `False`). |
| **del** | Ett kommando som används för att permanent radera en specifik nyckel och dess värde ur en dictionary (t.ex. `del pokedex["Pikachu"]`). |
| **update()** | En metod som används för att slå ihop två dictionaries, eller uppdatera värdena i en befintlig dictionary med ny data på en gång. |
