2. Classification accuracy

````python
# Inmatning
TP = 2   
FP = 2    
FN = 11   
TN = 985 

# Beräkning 
accuracy = (TP + TN) / (TP + TN + FP + FN)

if (TP + FP) > 0:
    precision = TP / (TP + FP)
else:
    precision = 0

if (TP + FN) > 0:
    recall = TP / (TP + FN)
else:
    recall = 0

# F1-score är det harmoniska medelvärdet av Precision och Recall
f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

# Skriv ut resultaten i procent (avrundat till 1 decimal)
print("--- Utvärdering av maskininlärningsmodellen ---")
print(f"Accuracy (Exakthet):  {accuracy * 100:.1f}%")
print(f"Precision (Precision): {precision * 100:.1f}%")
print(f"Recall (Känslighet):   {recall * 100:.1f}%")
print(f"F1-score:              {f1_score * 100:.1f}%")

print("\nSlutsats:")
if recall < 0.50:
    print("Modellen är livsfarlig! Den missar alldeles för många bränder (Lågt Recall).")
print(' "tog hjälp av AI för den här uppgiften" ')
````

### 7. Theory

#### a) What is the difference between input() and print() in Python?
* **`input()` tar emot data:** Den pausar programmet och väntar på att användaren ska skriva in något på tangentbordet.
* **`print()` skickar ut data:** Den visar text, tal eller resultat på skärmen för användaren.
* *Kort sagt:* `input()` är för **inmatning** (input) och `print()` är för **utmatning** (output).

#### b) What data type does the input() function return by default? How can you use it to get numeric input?
* **Standardtyp:** `input()` returnerar alltid en **sträng (`str`)**, det vill säga text. Även om användaren skriver siffran `5`, ser Python det som textsträngen `"5"`.
* **Numerisk inmatning:** För att göra om texten till ett tal måste du omvandla den (typkonvertera) med `int()` för heltal eller `float()` för decimaltal.

```python
# Exempel för decimaltal
cm = float(input("Mata in cm: "))
```

#### c) Explain how formatted string literals (f-strings) improve readability in output statements. Give an example.
* **Varför det är bättre:** f-strings gör koden mer lättläst eftersom du kan placera dina variabler direkt inuti textsträngen med måsvingar `{}`. Du slipper krångla med plustecken (`+`) eller kommatecken (`,`), vilket ger en renare och mer naturlig kodstruktur.

```python
namn = "Anna"
ålder = 25
# Lätt att läsa och förstå:
print(f"{namn} är {ålder} år gammal.")
```

#### d) Can a variable change its data type during runtime in Python? Show an example.
* **Svar:** **Ja**, en variabel kan fritt byta datatyp under programmets gång. 

```python
x = 10       # Här är x ett heltal (int)
x = "Kalle"  # Nu har x bytt till en textsträng (str) utan problem!
```

#### e) What does it mean that Python is "dynamically typed"? How is that different from statically typed languages?
* **Dynamiskt typtypat (Python):** Du behöver inte tala om för Python vilken datatyp en variabel har när du skapar den. Python listar ut det automatiskt i farten (vid runtime) baserat på värdet du sparar i den.
* **Statiskt typtypat (t.ex. Java eller C++):** Här *måste* du uttryckligen låsa variabeln till en viss datatyp när den skapas (t.ex. `int x = 10;`). Variabeln kan sedan aldrig hålla någon annan typ av data, vilket kontrolleras redan innan programmet körs (vid kompilering).

### Glossary

| Terminology | Explanation |
| :--- | :--- |
| **data type** | Bestämmer vilken sorts värde en variabel har (t.ex. text, heltal eller decimaltal) och vad man kan göra med det. |
| **variable** | En namngiven behållare i datorns minne som används för att spara och återanvända data i ett program. |
| **assignment** | Processen att tilldela (spara) ett värde till en variabel, vilket görs med ett likamedtecken (`=`). |
| **dynamically typed** | Betyder att programmeringsspråket automatiskt listar ut datatypen i farten, utan att du behöver skriva ut den i koden. |
| **input** | Den data eller information som skickas *in* till programmet, till exempel när en användare skriver något på tangentbordet. |
| **output** | Den data eller information som programmet skickar *ut*, till exempel text som visas på skärmen. |
| **type casting** | Att manuellt tvinga eller omvandla data från en datatyp till en annan (kallas även typkonvertering, t.ex. text till tal). |
| **boolean** | En datatyp som bara kan ha ett av två värden: sant (`True`) eller falskt (`False`). |
| **string** | En datatyp som består av en sekvens av tecken, det vill säga vanlig text inom citattecken. |
| **f-string** | Ett smidigt sätt i Python att formatera text där man kan stoppa in variabler direkt i en textsträng med måsvingar `{}`. |
| **indentation** | Indrag (mellanslag eller tab) i början av en kodrad. I Python används detta för att visa vilken kod som hör ihop i block (t.ex. i en if-sats). |
| **convention** | Gemensamma överenskommelser eller "best practices" inom programmering för hur man bör skriva ren och läsbar kod. |
